```python

import math
import fractions


# Menyamakan penyebut menggunakan KPK
def menyamakan_penyebut(
    p1: fractions.Fraction,
    p2: fractions.Fraction
):
    penyebut1 = p1.denominator
    penyebut2 = p2.denominator

    fpb = math.gcd(penyebut1, penyebut2)
    kpk = (penyebut1 * penyebut2) // fpb

    # Faktor pengali untuk masing-masing pecahan
    faktor1 = kpk // penyebut1
    faktor2 = kpk // penyebut2

    sulap1 = p1 * faktor1
    sulap2 = p2 * faktor2

    print("KPK penyebut:", kpk)
    print("Pecahan pertama:", sulap1)
    print("Pecahan kedua:", sulap2)


p1 = fractions.Fraction(1, 2)
p2 = fractions.Fraction(1, 3)

menyamakan_penyebut(p1, p2)
```
